Separating Research Data from the Real Patients Data

```python
import os
from dotenv import load_dotenv

# Loading the environment variables from the .env file
load_dotenv()

SECRET = os.environ.get("SECRET_KEY")
```

```python
import hashlib
import hmac

def research_patient_id(patient_id: int) -> str:
    digest = hmac.new(
        SECRET,
        str(patient_id).encode(),
        hashlib.sha256
    ).hexdigest()[:12]

    return f"PAT_{digest.upper()}"

def research_report_id(report_id: int) -> str:
    digest= hmac.new(
        SECRET,
        str(report_id).encode(),
        hashlib.sha256
    ).hexdigest()[:12]
    return f"REP_{digest.upper()}"
```
